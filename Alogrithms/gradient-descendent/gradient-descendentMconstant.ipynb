```python
from sklearn.datasets import make_regression
X,y = make_regression(n_samples=4, n_features=1, n_informative=1, n_targets=1,noise=80,random_state=13)

X,y
```

```python
import matplotlib.pyplot as plt

plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
     

reg = LinearRegression()
reg.fit(X,y)
     
```

```python



reg.intercept_
     


```

```python
#m 
reg.coef_
     
```

```python

plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red')

```

```python
# Lets apply Gradient Descent assuming slope is constant m = 78.35
# and let's assume the starting value for intercept b = 0
b=100
m=78.350
y_pred=((m*X)+ b).reshape(4)

```

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred,color='#00a65a',label='b = 0')
plt.legend()
plt.show()
     
```

```python
#now apply gradient descendent with 0.001 learning rate and the slope will be dy/dx=-2(y-m*X.ravel()-b) from MSE loss funtion
#eta=lr(learning Rate=0.001)
#bnew=bold-eta*slope
import numpy as np
b=100
m=78.350
slope=-2*np.sum(y-m*X.ravel()-b)


slope
```

```python
lr = 0.1

step_size = slope*lr
step_size
```

```python
# b new is

bnew=b-step_size
```

```python

y_pred1 = ((78.35 * X) + b).reshape(4)

plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred1,color="#000590",label='b = {}'.format(b))
plt.plot(X,y_pred,color="#FCFCFC",label='b = 0')
plt.legend()
plt.show()
```

```python

# Iteration 2
loss_slope = -2 * np.sum(y - m*X.ravel() - b)
loss_slope
     
```

```python

step_size = loss_slope*lr
step_size
```

```python

b = b - step_size
b
     
```

```python

y_pred2 = ((78.35 * X) + b).reshape(4)

plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred2,color='#00a65a',label='b = {}'.format(b))
plt.plot(X,y_pred1,color="#505B59",label='b = {}'.format(b))
plt.plot(X,y_pred,color='#A3E4D7',label='b = 0')
plt.legend()
plt.show()
```

```python

# Iteration 3
loss_slope = -2 * np.sum(y - m*X.ravel() - b)
loss_slope


step_size = loss_slope*lr
step_size
     

b = b - step_size
b
     

y_pred3 = ((78.35 * X) + b).reshape(4)

plt.figure(figsize=(15,15))
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred3,color='#00a65a',label='b = {}'.format(b))
plt.plot(X,y_pred2,color='#A3E4D7',label='b = {}'.format(b))
plt.plot(X,y_pred1,color='#A3E4D7',label='b = {}'.format(b))
plt.plot(X,y_pred,color='#A3E4D7',label='b = 0')
plt.legend()
plt.show()
     
```
